## 📌 Visão Geral do Projeto

### Problem Statement
A Síndrome Respiratória Aguda Grave (SRAG) é uma condição de saúde pública monitorada pelo Ministério da Saúde do Brasil. Identificar, no momento da internação, quais pacientes têm maior risco de óbito pode permitir priorização de recursos clínicos e intervenção precoce. O dataset SRAG 2025 reúne ~335 mil notificações hospitalares com variáveis clínicas e demográficas. O problema é de classificação binária desbalanceada (apenas ~7,5% dos desfechos são óbitos), o que exige estratégias especiais de balanceamento e escolha de métricas adequadas.

### Objetivos
- **Principal:** Desenvolver um modelo de classificação para prever o risco de óbito hospitalar por SRAG com ROC-AUC > 0,88 e F1-Score ≥ 0,60 na classe de óbito
- **Secundários:**
  - Identificar os principais fatores clínicos e demográficos associados ao óbito por SRAG
  - Avaliar e comparar diferentes estratégias de balanceamento de classes (RUS, SMOTE, SMOTEENN)
  - Otimizar o threshold de classificação para maximizar o Recall (triagem clínica)

### 📊 Trabalho Realizado

#### O que fizemos:
Pedro avaliou as três estratégias de balanceamento (RUS, SMOTE, SMOTEENN) de forma sistemática e concluiu que XGBoost + RUS obteve o melhor ROC-AUC (~0,91) com muito menos custo computacional que SMOTE (que gerava 447K amostras sintéticas). Corrigimos o data leakage do pipeline anterior: adotamos a divisão 70/15/15 (treino/validação/teste), onde o threshold é escolhido no conjunto de validação e a avaliação final ocorre apenas no conjunto de teste. Guilherme analisou a relevância das colunas via feature importance do XGBoost — `SUPORT_VEN`, `NU_IDADE_N` e `UTI` foram as mais importantes. Pedro definiu PR-AUC e Recall como métricas prioritárias, dado o desequilíbrio das classes e o contexto clínico (falso negativo é mais custoso que falso positivo). Leo melhorou a organização e documentação do notebook.

### 🚧 Desafios e Soluções

| Desafio | Solução Adotada | Status |
|---------|-----------------|--------|
| Data leakage: threshold escolhido no conjunto de teste nos experimentos anteriores | Adotar divisão 70/15/15 com threshold otimizado na validação e avaliação final apenas no teste | ✅ Resolvido |
| F1-Score abaixo de 0,60 mesmo com balanceamento | Priorizar PR-AUC e Recall como métricas principais; threshold dinâmico via curva PR | ✅ Resolvido |

### 📝 Decisões Importantes
1. **Divisão 70/15/15 e threshold na validação:** Para garantir avaliação honesta do modelo, o conjunto de teste permanece intocado até a avaliação final. O threshold é escolhido via curva Precision-Recall no conjunto de validação.
2. **PR-AUC como métrica principal de comparação:** Em bases desbalanceadas, ROC-AUC pode ser otimista. PR-AUC foca no desempenho na classe minoritária (óbito) e é mais informativa neste cenário clínico.

### 🔍 Principais Aprendizados

#### Técnico
1. Em bases desbalanceadas, a acurácia é uma métrica enganosa — PR-AUC e Recall são muito mais informativas para a classe minoritária.
2. O threshold padrão de 0,5 não é ideal: a otimização via curva Precision-Recall na validação permite encontrar pontos de corte mais adequados ao contexto de uso.
3. Random Undersampling (RUS) produziu resultados equivalentes ao SMOTE com fração do custo computacional — em datasets grandes, simplicidade vence.

#### Trabalho em Equipe
1. Dividir responsabilidades por etapas do pipeline (dados, modelagem, documentação) evitou retrabalho e acelerou o desenvolvimento.
2. O feedback iterativo do professor ao longo das sprints foi fundamental para corrigir problemas metodológicos (data leakage, separação de notebooks) antes da entrega final.
